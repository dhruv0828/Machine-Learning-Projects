**SECTION 4 – TRAFFIC & POLLUTION FORECASTING**

```python
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA

from sklearn.metrics import mean_absolute_error
```

```python
df = pd.read_csv("urban_traffic_pollution_dataset.csv")
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df.set_index('date', inplace=True)

print(df.head())
```

Time-Based Structure

 Daily average traffic and pollution

```python
daily_data = df.resample('D').mean(numeric_only=True)

print(daily_data.head())
```

Visualization

Traffic Density over time

```python
plt.figure(figsize=(10,5))

plt.plot(daily_data['traffic_density'])

plt.title("Traffic Density Over Time")
plt.xlabel("Date")
plt.ylabel("Traffic Density")

plt.show()
```

PM25 Pollution over time

```python
plt.figure(figsize=(10,5))

plt.plot(daily_data['pm25'])

plt.title("PM25 Pollution Over Time")
plt.xlabel("Date")
plt.ylabel("PM25")

plt.show()
```

Observations:
 - Peak traffic hours may appear
 - Pollution rises during heavy traffic
 - Recurring daily patterns may exist

Check Data Stability

```python
result = adfuller(daily_data['pm25'].dropna())

print("ADF Statistic:", result[0])
print("p-value:", result[1])
```

If p-value > 0.05:



  Data is unstable (non-stationary)

Handle Instability

Apply differencing if needed

```python
daily_data['pm25_diff'] = daily_data['pm25'].diff()
```

Model Selection Factors

Factors:
 1. Trend in data
 2. Seasonality
 3. Data size
 4. Forecast accuracy
 5. Computational complexity

ARIMA works well for time-series forecasting.

Train Forecasting Model

Remove missing values

```python
series = daily_data['pm25'].dropna()
```

Split data

```python
train = series[:-7]
test = series[-7:]
```

Build ARIMA model

```python
model = ARIMA(train, order=(1,1,1))

model_fit = model.fit()
```

Forecast

```python
forecast = model_fit.forecast(steps=7)

print("\nFORECAST VALUES")
print(forecast)
```

Model Evaluation

```python
mae = mean_absolute_error(test, forecast)

print("\nMAE:", mae)
```

 Lower MAE means better forecasting accuracy.

Future Prediction

```python
future_forecast = model_fit.forecast(steps=5)

print("\nNEXT 5 DAYS PREDICTION")
print(future_forecast)
```

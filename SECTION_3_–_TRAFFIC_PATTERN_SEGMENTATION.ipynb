**SECTION 3 – TRAFFIC PATTERN SEGMENTATION**

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("urban_traffic_pollution_dataset.csv")
```

Selected Features

```python
features = [
    'traffic_density',
    'car_count',
    'bus_count',
    'truck_count',
    'bike_count',
    'temperature',
    'humidity',
    'wind_speed'
]
```

These features help identify:
 - Traffic volume
 - Vehicle composition
 - Weather impact on traffic

Data Preprocessing

  Convert columns to numeric

```python
for col in features:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

Fill missing values

```python
df.fillna(df.median(numeric_only=True), inplace=True)
```

Feature Scaling

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(df[features])
```

Reduce Complexity using PCA

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Reduced Shape:", X_pca.shape)
```

CA keeps important information while reducing dimensions.

Find Optimal Number of Clusters

```python
wcss = []

for i in range(1, 11):

    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    wcss.append(kmeans.inertia_)
```

Elbow Method

```python
plt.plot(range(1, 11), wcss, marker='o')

plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.show()
```

Apply K-Means Clustering

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df['Cluster'] = kmeans.fit_predict(X_scaled)

print(df[['traffic_density', 'Cluster']].head())
```

Cluster Interpretation

```python
cluster_summary = df.groupby('Cluster')[features].mean()

print("\nCLUSTER SUMMARY")
print(cluster_summary
```

Example Interpretation:
* Cluster 0 -> High traffic areas
* Cluster 1 -> Moderate traffic areas
* Cluster 2 -> Low traffic areas
